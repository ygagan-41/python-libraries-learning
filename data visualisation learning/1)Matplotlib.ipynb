```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
y = x**2
plt.plot(x,y)
```

```python
plt.title("Graph of X and Y")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.plot(x,y)
```

subplots in matplotlib


```python
import matplotlib.pyplot as plt
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x**2,y**2)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x,y)
```

Object Oriented Way

```python
fig = plt.figure(figsize=(6,8),dpi=100)
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.5,0.8,0.8])
axis2.plot(x,y)
```

TYpes of plot

Savings PLot

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
