```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
df = sns.load_dataset('tips')
```

```python
df
```

```python
sns.countplot(x = df['sex'])
```

```python
sns.countplot(x = df['sex'] , hue = df['smoker'])
```

```python
sns.barplot(x = df['sex'] , y = df['total_bill'])
```

```python
sns.barplot(x = df['sex'] , y = df['total_bill'], estimator= np.sum)
```

```python
sns.boxplot(x = 'tip' , y = 'total_bill' ,data=df,palette='rainbow')
```

```python
sns.violinplot(x = 'sex' , y= 'tip',data=df)
```

```python
sns.stripplot(x='tip',y='total_bill',data=df)
```
