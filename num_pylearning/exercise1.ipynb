```python
import numpy as np
```

```python
#valid soduku ques
s = np.array([
      [5, 3, 4, 6, 7, 8, 9, 1, 2], #in soduku evry row and column sum is 45
      [6, 7, 2, 1, 9, 5, 3, 4, 8],
      [1, 9, 8, 3, 4, 2, 5, 6, 7],

      [8, 5, 9, 7, 6, 1, 4, 2, 3],
      [4, 2, 6, 8, 5, 3, 7, 9, 1],
      [7, 1, 3, 9, 2, 4, 8, 5, 6],

      [9, 6, 1, 5, 3, 7, 2, 8, 4],
      [2, 8, 7, 4, 1, 9, 6, 3, 5],
      [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
s.sum()
```

```python
np.sum(s,axis=0) #columns sum
```

```python
np.sum(s,axis=1) #rows sum
```

```python
#check if rows == 45 or not
c = np.sum(s,axis = 1)
c == 45
```

```python
c = np.sum(s,axis = 1)
for i in c:
    if i != 45:
        print("soduku is not valid")
        break
else:
    print("for rows it is valid")
```

```python
b = np.sum(s,axis = 0)
for i in b:
    if i != 45:
        print("soduku is not valid")
        break
else:
    print("for columns it is valid")
```
