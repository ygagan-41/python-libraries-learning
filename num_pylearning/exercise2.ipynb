```python
import numpy as np
data = np.array([
    [18, 85, 78],   #student1
    [19, 92, 88],   #student2
    [17, 76, 95],   #student3
    [18, 65, 70],   #student4
    [20, 90, 85]    #student5
])
```

```python
#get the shape of the matrix
data.shape
```

```python
#find the average age of the students
np.mean(data[:,0])
```

```python
#extract math marks of all students
data[:,1:2]
```

```python
#get details of the student who score more than 90 in math
data[data[:,1] > 90]
```

```python
#find the highest mark of science 
np.max(data[:,2])
```

```python
#increase math marks of all students by 5
data = np.array([
    [18, 85, 78],   #student1
    [19, 92, 88],   #student2
    [17, 76, 95],   #student3
    [18, 65, 70],   #student4
    [20, 90, 85]    #student5
])
data[:,1] += 5
data
```

```python
#find how many students are younger than 19
len(data[data[:,0] < 19])
```

```python
#calculate the average marks in each subject ( column-wise mean).
np.mean(data[:,1:],axis=0)
```

```python
#get the data of students who scored at least 80 in both subjects.
data[(data[:,1] >= 80) & (data[:,2] >= 80)]
```
