```python
#reshaping and resizing
import numpy as np
arr = np.arange(20)
arr
```

```python
arr.reshape(5,4) #reshape
```

```python
arr.resize(4,5) #resize
arr
```

```python
#slicing and indexing
arr = np.arange(10)
arr
```

```python
arr[3:6] #slicing
```

```python
arr[5] #indexing
```

```python
#for 2d matrix 
arr = np.arange(9).reshape(3,3)
arr
```

```python
arr[0:2,0:1] #slicing
```

```python
arr[0,2] #indexing
```

```python
#boolean indexing
arr = np.arange(9).reshape(3,3)
arr
```

```python
bool_index = arr%2 == 0
bool_index
```

```python
#broadcasting
lst = [1,2,3,4,5]
arr = np.array(lst)
arr + 10
```

```python
#shallowcopy and deepcopy
arr = np.arange(10)
arr
```

```python
slice = arr[0:5] #in shallow copy , changes did in other array doesn't affect original array
slice
slice * 10
arr
```

```python
arr = np.arange(10) # in deep copy , changes affect in original array
brr = arr
brr[0] = 10
brr
arr
```
