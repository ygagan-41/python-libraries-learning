```python
#matrix operation
import numpy as np
arr = np.array([[1,2],[1,2]])
brr = np.array([[7,2,3],[5,2,3]])
np.dot(arr,brr) #or arr@brr
```

```python
#advance array manipulation
arr = np.array([1,2,3,4]) #matrix stack nhi hosakti
brr = np.array([5,3,3,4])
np.vstack((arr,brr))
```

```python
np.hstack((arr,brr))
```

```python
np.column_stack((arr,brr))
```
