```python
import numpy as np
import pandas as pd
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
#finding missing data
df.isna() #bata dega kon kon si value null hai 
```

```python
df.isna().sum() #bata dega kis column me kitni value null hai
```

```python
df.isna().any() #bata dega ki column koi null value exist karti hai ya nhi
```

```python
#removing missing data
df.dropna() #row jisme null value nhi hai
```

```python
df.dropna(thresh=3) #row jisme atleast 3 value null h
```

```python
#Filling the missing data
df.fillna(0) #jitni bhi null value hogi usme null bharjaega
```

```python
values = {'A':100,'B':20,'C':40,'D':200}
df.fillna(value=values) #null value ko kuch values dena
```

```python
df.fillna(df.mean())
```
