```python
import pandas as pd
data = {
   'Name': ['John', 'Anna', 'Peter', 'Linda'],
   'Age': [28, 34, 29, 42],
   'City': ['New York', 'Paris', 'Berlin', 'London'],
   'Salary': [65000, 70000, 62000, 85000]
}
a = pd.DataFrame(data)
a
```

```python
#selection and indexing of columns
a['Name'] #indexing selection
```

```python
a[["Name","City"]]
```

```python
#creating a new columns

a["dept"] =["cse","finance","hr","cse"]
a
```

```python
#remove columns
a.drop("dept",axis=1) #column will removed but not from original array
```

```python
a.drop("dept", axis=1, inplace=True) #remove column from original array
```
